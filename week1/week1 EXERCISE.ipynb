To demonstrate your familiarity with OpenAI API, and also Ollama, build a tool that takes a technical question,  
and responds with an explanation. This is a tool that you will be able to use yourself during the course!

```python
# imports
import os
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv(override=True)
openai_api_key = os.getenv('OPENAI_API_KEY')
ollama_api_key = 'ollama'
```

```python
# constants
MODEL_GPT = 'gpt-4o-mini'
MODEL_LLAMA = 'llama3.2'
```

```python
# set up environment
OPENAI_BASE_URL = "https://api.openai.com/v1"
OLLAMA_BASE_URL = "http://localhost:11434/v1"

def explain_code(url, model, api_key, messages):
    openai = OpenAI(base_url=url, api_key=api_key)
    response = openai.chat.completions.create(
        model=model,
        messages=messages,
        stream=True
    )
    content = ""
    for chunk in response:
        if chunk.choices[0].delta.content:
            print(chunk.choices[0].delta.content, end="")
            content += chunk.choices[0].delta.content
    return content
```

```python
# here is the question; type over this to ask something new

question = """
Please explain what this code does and why:
yield from {book.get("author") for book in books if book.get("author")}
"""

messages = [
    {"role": "system", "content": "You are a seasoned software engineer."},
    {"role": "user", "content": question}
]
```

```python
# Get gpt-4o-mini to answer, with streaming
response = explain_code(OPENAI_BASE_URL, MODEL_GPT, openai_api_key, messages)
```

```python
# Get Llama 3.2 to answer
response = explain_code(OLLAMA_BASE_URL, MODEL_LLAMA, ollama_api_key, messages)
```
